```python
import os
import sys
import torch
import numpy as np
import json
from pathlib import Path
import subprocess
from datetime import datetime
import matplotlib.pyplot as plt
#import cv2
```

```python
MODELS_DIR = "models\visual\detectron2"

def run_coco_finetuning(config_path, output_base="output"):
    """Run COCO fine-tuning phase"""
    
    from models.visual.detectron2.training.train_coco import train_coco_model
    
    timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
    output_dir = Path(output_base) / f"coco_finetune_{timestamp}"
    
    print(f"Starting COCO fine-tuning...")
    print(f"Output directory: {output_dir}")
    
    # Train
    result_dir = train_coco_model(
        config_path,
        str(output_dir),
        resume=False,
        gpu_id=0
    )
    
    return result_dir
# Check if we should run COCO fine-tuning
run_coco = input("Run COCO fine-tuning? (y/n): ").lower().strip() == 'y'

if run_coco:
    config_path = "models\visual\detectron2\config\coco_pretrain.yaml"
    if config_path:
        coco_output = run_coco_finetuning(str(config_path))
        print(f"COCO fine-tuning complete. Model saved at: {coco_output}")
        
        # Update lecture config to use COCO-finetuned weights
        coco_weights = Path(coco_output) / "model_final.pth"
        if coco_weights.exists():
            print(f"COCO weights available at: {coco_weights}")
    else:
        print(f"COCO config not found at {config_path}")
        print("Using default COCO pre-trained weights.")
else:
    print("Skipping COCO fine-tuning. Using default COCO pre-trained weights.")
```
